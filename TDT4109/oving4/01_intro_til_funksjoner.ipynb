## Innføring - Del 1: Enkle funksjoner uten parametre og returverdi

Dette er ikke en del av oppgaven, men kan være lurt å lese før du går videre.

Programmet nedenfor tenkes å skulle skrive ut en masse data (hvorav bare fragment er vist her).

Mellom hver bolk av data vil vi ha et skille med blank linje, horisontal strek og en ny blank linje.

Dette gir identisk kode alle stedene markert med `##`, som er unødig tungvint.

```python
# EKSEMPEL UTEN FUNKSJON
print("Eksempel der vi BURDE brukt en funksjon")
print()  ##
print("---------------------")  ##
print()  ##
print("Data fra måler 1:", 23.4, 21.2, 21.9)
print("Osv., ...")
print()  ##
print("---------------------")  ##
print()  ##
print("Data fra måler 2: ...")
print()  ##
print("---------------------")  ##
print()  ##
print("Data fra måler 3: ...")
print()  ##
print("---------------------")  ##
print()  ##
```

Nedenfor er det løst bedre:

- Kodesnutten som trengs flere steder i programmet, defineres øverst som en funksjon, her med navnet `skille()`
- Trenger da kun skrive `skille()` de stedene i koden hvor vi ønsker disse tre `print`-setningene utført.

```python
# EKSEMPEL MED FUNKSJON


def skille():
    print()
    print("---------------------")
    print()


print("Samme eksempel MED funksjon")
skille()
print("Data fra måler 1:", 23.4, 21.2, 21.9)
print("Osv., ...")
skille()
print("Data fra måler 2:", 18.4, 19.1, 18.7)
skille()
print("Data fra måler 3:", 19.9, 19.1, 18.2)
skille()
```

Utførelsen av denne koden starter på toppen av **scriptet**, det vil si koden som som er ute på vanlig marg.

Det første som printes er derfor "Samme eksempel MED funksjon".

Deretter de tre printene i skille-funksjonen.

Deretter `"Data fra måling 1: ..."` osv. Alt i alt blir resultatet dermed identisk med koden der vi ikke hadde funksjon, men koden med funksjon er kortere og ryddigere fordi vi unngår å gjenta identiske kodelinjer flere ganger.

### a)

Endre koden på følgende måte:

- Definer en funksjon `siksak()` som utfører de fire linjene som er merket med `##`.

Utskriften etter endringen din skal være eksakt den samme som før programmet ble endret. Hvis du er i tvil om hva du skal gjøre her, les tutorial ovenfor som viser et lignende eksempel.

## Innføring - Del 2: Funksjoner med parametre, men fortsatt uten returverdi

Dette er ikke en del av oppgaven, men kan være lurt å lese før du går videre.

[Innføring - Del $1$](#innføring---del-1-enkle-funksjoner-uten-parametre-og-returverdi) og oppgave a) illustrerte hovedpoenget med funksjoner: At vi har lik kode flere sted i et program, denne kan da defineres en gang for alle og deretter kalles ved navn, så slipper vi å gjenta kodelinjer, og programmet blir ryddigere.

Uten parametre har funksjoner likevel begrenset slagkraft. En funksjon uten parametre gjør eksakt det samme hver gang vi kaller den. Men ofte er behovet å gjøre nesten det samme, men ikke helt. En kvadratrotfunksjon trenger for eksempel å kunne regne ut roten av ulike tall - en funksjon som bare kan regne roten av $2$, er ikke særlig nyttig. Nedenfor vises et eksempel som ligner på eksemplet fra oppgave a). Vi har en utskrift hvor det inngår overskrifter i hashtag-bokser. Men koden er bare nesten lik, siden overskriftens tekst endrer seg fra gang til gang. Hvis vi lager en funksjon som alltid skriver INTRODUKSJON, kan denne ikke brukes for de neste overskriftene.

```python
# EKSEMPEL UTEN FUNKSJON
print()
print("################")
print("# INTRODUKSJON #")
print("################")
print()
print("her kommer noen innledende forklaringer...")
print()
print("##########################")
print("# DATA FRA EKSPERIMENT 1 #")
print("##########################")
print()
print("Div. data:", 123, 345, 432, 356)
print()
print("##########################")
print("# DATA FRA EKSPERIMENT 2 #")
print("##########################")
print()
print("Osv...")
```

Løsningen blir å utstyre funksjonen med en **parameter**, som gir funksjonen fleksibilitet til å endre litt oppførsel fra gang til gang:

```python
# SAMME EKSEMPEL MED FUNKSJON


def box_heading(tekst):
    bredde = len(tekst) + 4
    print()
    print("#" * bredde)
    print(f"# {tekst} #")
    print("#" * bredde)
    print()


box_heading("INTRODUKSJON")
print("her kommer noen innledende forklaringer...")
box_heading("DATA FRA EKSPERIMENT 1")
print("Div. data:", 123, 345, 432, 356)
box_heading("DATA FRA EKSPERIMENT 2")
print("Osv...")
```

Programmet starter med å utføre `box_heading("INTRODUKSJON")`. Dette gjør at vi kjører funksjonen `box_heading`, med argumentet `"INTRODUKSJON"` byttet inn for parameteren `tekst`.

Setningen `print(f"# {tekst} #")` gjør da at denne strengen skrives midt i boksen.

Funksjonen `len(tekst)` gir oss lengden på strengen, dette gjør det mulig for oss å tilpasse bredden på boksen til hva som gis inn for parameteren.

I neste kall av `box_heading` er det derimot argumentet `"DATA FRA EKSPERIMENT 1"` som byttes inn for parameteren `tekst`. Osv. Bruk av parameter gjør det dermed mulig for funksjonen `box_heading` å skrive ulik tekst i boksen fra gang til gang.

## Innføring - Del 3: Funksjoner med parametre *og* returverdi

Dette er ikke en del av oppgaven, men kan være lurt å lese før du går videre

I oppgave a) og b) har vi sett på funksjoner uten returverdi. Dette kan være greit for funksjoner som spesifikt skal skrive eller tegne på skjermen.

Men for funksjoner som skal beregne en verdi (og svært mange funksjoner skal nettopp det), trenger vi å få ut et resultat av beregningen. Til dette brukes **returverdi**.

Det trenger ikke å være en matematisk beregning. For å vise noe annet, gis under et eksempel relatert til språk, nemlig en funksjon for å gradbøye adjektiv.

Ideen med funksjonen `komparativ()` er at den får inn et adjektiv i vanlig positiv form, og da kan bøye til komparativ, for eksempel snill -> snillere, interessant -> mer interessant.

For at ikke koden skal bli for lang, er eksemplet grovt forenklet - mange adjektiv vil bli bøyd feil.

Å gi returverdi fra en funksjon, oppnås med ordet `return` (linje $7$). Verdien av det som står etter return (her: variabelen svar, som vil være en streng, for eksempel "snillere") vil bli returnert fra funksjonen.

Med "snillere" som returverdi vil da `komparativ(adj)` i linje $11$ printe "Hah! Jeg er mye snillere !"

Lenger nede i programmet, linje $16$, vil bøyningen (for eksempel "snillere") bli satt inn i variabelen fasit, som deretter brukes i en `if`-test.

```python
def komparativ(adj):
    # GROVT FORENKLET FOR Å KUNNE FOKUSERE PÅ HOVEDPOENGET
    if len(adj) >= 8:  # Unøyaktig
        return "mer " + adj
    return adj + "ere"


### TILLEGG 1, ny funksjon, kommer her:


# DEL AV KODEN HVOR SYSTEMET DISSER BRUKEREN
adj = input("Beskriv deg selv med et adjektiv: ")
print(f"Hah! Jeg er mye {komparativ(adj)}!")

### TILLEGG 2 kommer her ...


# DEL AV KODEN HVOR BRUKEREN TRENER PÅ Å GRADBØYE
adj = input("Skriv inn et adjektiv: ")
svar = input(f"Hva er komparativ for {adj}? ")
fasit = komparativ(adj)

if svar == fasit:
    print("Korrekt!")
else:
    print(f"Feil, det var {fasit}.")

### TILLEGG 3 kommer her...
```

Men hva oppnår vi med returverdi her? Hvorfor ikke like godt bruke `print()` i slutten av funksjonen, heller enn å returnere en verdi som printes senere??

**Med returverdi oppnår vi at funksjonen blir mer generelt anvendelig.** Funksjonen lar seg bruke for to ulike behov i dette lille programmet:

- Bruk $1$, linje $11$, her er det programmet som autogenererer komparativ form og printer ut på skjermen
- Bruk $2$, linje $16$, her er det brukeren som prøver å svare hva komparativ er, programmet tester med en `if`-setning om det var korrekt

**Hadde vi derimot printet i slutten av funksjonen, ville anvendeligheten har vært mindre.**

- Printe "Hah! Jeg er ..." i slutten av funksjonen? Da passer den bare for bruk $1$.
- Gjøre en `if`-test i slutten av funksjonen og printe Korrekt / Feil? Da passer den bare for bruk $2$.
- Gjøre bare print(svar) i slutten av funksjonen istedenfor return svar? Da passer den verken for bruk $1$ eller bruk $2$.

På samme måte ville input i starten av funksjonen (istedenfor parameteren adj) ødelegge den generelle anvendeligheten.

- `input("Beskriv deg selv...")` i funksjonen - og den passer kun for bruk $1$.
- `input("Skriv inn...")` osv. i funksjonen - og den passer kun for bruk $2$.

For å lage generelt anvendelige funksjoner er det vanligvis lurt å unngå input og print i funksjonen - annet enn for funksjoner som spesifikt har som mål akkurat å kommunisere med brukeren.

### b)

Ta utgangspunkt i koden nedenfor, som også er forklart like over. Kjør koden og sjekk at den virker. Ikke la deg bekymre deg om at en del adjektiv bøyes feilaktig.

```python
def komparativ(adj):
    # GROVT FORENKLET FOR Å KUNNE FOKUSERE PÅ HOVEDPOENGET
    if len(adj) >= 8:  # Unøyaktig
        return "mer " + adj
    return adj + "ere"


def superlativ(adj):
    # GROVT FORENKLET, SAMME PRINSIPP SOM komparativ()
    if len(adj) >= 8:  # Unøyaktig
        return "mest " + adj
    return adj + "est"


# DEL AV KODEN HVOR SYSTEMET DISSER BRUKEREN
adj = input("Beskriv deg selv med et adjektiv: ")
print(f"Hah! Jeg er mye {komparativ(adj)}!")
print(f"Jeg er faktisk {superlativ(adj)} i hele verden!")


# DEL AV KODEN HVOR BRUKEREN TRENER PÅ Å GRADBØYE
adj = input("Skriv inn et adjektiv: ")
svar = input(f"Hva er komparativ for {adj}? ")
fasit = komparativ(adj)

if svar == fasit:
    print("Korrekt!")
else:
    print(f"Feil, det var {fasit}.")

svar = input(f"Hva er superlativ for {adj}? ")
fasit = superlativ(adj)

if svar == fasit:
    print("Korrekt!")
else:
    print(f"Feil, det var {fasit}.")
```

**Oppgave:** Lag en tilsvarende funksjon `superlativ(adj)` som tar inn som parameter et adjektiv i vanlig positiv form, bøyer det til superlativ (for eksempel snill -> snillest, interessant -> mest interessant), og returnerer superlativformen.

- Der det står `### TILLEGG 3`, legg til en programdel som spør brukeren "Hva er superlativ for ..." (det adjektivet som ble skrevet inn...) og sjekker dette mot fasit på samme måte som koden nå gjør for komparativ.

## Innføring - Del 4: `main`-funksjonen

Dette er ikke en del av oppgaven, men kan være lurt å lese før du går videre.

Det er vanlig for et program å ha en `main`-funksjon som blir kalt når programmet starter. Det er denne funksjonen som kaller programmets andre funksjoner når de trengs.

Et eksempel på et program med en `main`-funksjon er vist under. Her er det oppdelt slik at funksjonen `kvadrattall(x)` gjør en beregning, mens `main()` står for kommunikasjon med brukeren.

I større program kunne det også ha vært egne funksjoner for å kommunisere med brukeren, slik at `main()` bare hadde gjort den overordnede styringen både av beregning og brukerkommunikasjon ved å kalle andre funksjoner.

Når kode som vi ellers ville hatt i scriptet puttes i en `main`-funksjon, blir scriptet kun én linje som kaller `main`-funksjonen, som vist nederst her. Kjør koden selv og se om du skjønner hvordan den funker:

```python
def kvadrattall(x):
    for y in range(1, x + 1):
        if y**2 == x:
            return y
    return False


def main():
    tall = int(input("Gi et heltall >0, eller 0 for å slutte: "))

    while tall > 0:
        rot = kvadrattall(tall)

        if rot:
            print(f"{tall} = {rot} * {rot}; m.a.o. et kvadrattall")
        else:
            print(f"{tall} er ikke et kvadrattall.")

        tall = int(input("Gi et heltall >0, eller 0 for å slutte: "))


# SCRIPT
if __name__ == "__main__":
    main()
```

```python
def komparativ(adj):
    # GROVT FORENKLET FOR Å KUNNE FOKUSERE PÅ HOVEDPOENGET
    if len(adj) >= 8:  # Unøyaktig
        return "mer " + adj
    return adj + "ere"


def superlativ(adj):
    # GROVT FORENKLET, SAMME PRINSIPP SOM komparativ()
    if len(adj) >= 8:  # Unøyaktig
        return "mest " + adj
    return adj + "est"


def sjekk_svar(svar, fasit):
    if svar == fasit:
        print("Korrekt!")
    else:
        print(f"Feil, det var {fasit}.")


def main():
    # DEL AV KODEN HVOR SYSTEMET DISSER BRUKEREN
    adj = input("Beskriv deg selv med et adjektiv: ")
    print(f"Hah! Jeg er mye {komparativ(adj)}!")
    print(f"Jeg er faktisk {superlativ(adj)} i hele verden!")

    # DEL AV KODEN HVOR BRUKEREN TRENER PÅ Å GRADBØYE
    adj = input("Skriv inn et adjektiv: ")
    sjekk_svar(input(f"Hva er komparativ for {adj}? "), komparativ(adj))
    sjekk_svar(input(f"Hva er superlativ for {adj}? "), superlativ(adj))


main()
```

## d)

Det er ofte vanlig å isolere et kall til `main`-funksjonen i en `if __name__ == "__main__":`-blokk, slik som vist under.

```python
if __name__ == "__main__":
    main()
```

Hvorfor er det vanlig å lage en `main`-funksjon i et program, istedenfor å bare la koden stå i scriptet? Og hvorfor er det vanlig å isolere kallet til `main` i en `if __name__ == "__main__":`-blokk? Skriv svaret ditt i tekstboksen under.

**Svar:** Med en `main`-funksjon samles programmets hovedflyt på ett sted, og variablene som brukes i flyten blir lokale i stedet for globale. Det gir ryddigere kode, færre utilsiktede sideeffekter og et tydelig inngangspunkt som er lett å finne.

`if __name__ == "__main__":` sørger for at `main()` bare kjøres når filen startes direkte som program. Blir filen importert fra en annen fil (for eksempel for å gjenbruke `komparativ` eller for å teste den), settes `__name__` til modulnavnet, og programmet starter da ikke og spør om input.
